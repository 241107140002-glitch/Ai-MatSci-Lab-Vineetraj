```python
from mp_api.client import MPRester

api_key = "yPl1CA8uMYXkTiKbFuH8f1w0oli668DJ"

with MPRester(api_key) as mpr:
    docs_bg = mpr.materials.summary.search(
        formula="ABO3",
        fields=["material_id", "formula_pretty", "band_gap"]
    )

    for doc in docs_bg:
        print("Material ID:", doc.material_id)
        print("Formula:", doc.formula_pretty)
        print("Band Gap:", doc.band_gap)
        print("--------------------")
```

```python
# Query 2: Formation energy and lattice volume

docs_thermo = mpr.materials.summary.search(
    formula="ABC3",
    fields=["material_id", "formation_energy_per_atom", "volume"]
)

for doc in docs_thermo:
    print("Material ID:", doc.material_id)
    print("Formation Energy per Atom:", doc.formation_energy_per_atom)
    print("Volume:", doc.volume)
    print("-" * 40)
```

```python
# 4. Convert to Pandas DataFrames
df_bg = pd.DataFrame([{
    "Material_ID": d.material_id, 
    "Formula": d.formula_pretty, 
    "Bandgap": d.band_gap
} for d in docs_bg])
```

```python
df_thermo = pd.DataFrame([{
    "Material_ID": d.material_id, 
    "Formation_Energy": d.formation_energy_per_atom, 
    "Volume": d.volume
} for d in docs_thermo])
```

```python
# 5. Merge DataFrames on the common Material ID
df_merged = pd.merge(df_bg, df_thermo, on="Material_ID")

# 6. Export to CSV
df_merged.to_csv("perovskites_data.csv", index=False)
print(f"Success! Exported {len(df_merged)} perovskites to CSV.")
```
